```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Generate 1000 random numbers uniformly distributed between 0 and 1
data = np.random.uniform(0, 1, 1000)
```

```python
# Create a histogram with 100 bins
plt.figure(figsize=(8, 5))
plt.hist(data, bins=100, color='skyblue', edgecolor='black')

# Label axes and title
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram of Uniformly Distributed Random Numbers")

# Save the figure as a PDF file
plt.savefig("uniform_random_histogram.pdf", format='pdf')

# Display the plot
plt.show()
```
