```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Create an array x ranging from 0 to 1 with 100 evenly spaced values
x = np.linspace(0, 1, 100)
```

```python
# Define a function that returns exp(x)
def my_exp(x):
    return np.exp(x)
```

```python
# Use the function to compute y = exp(x)
y = my_exp(x)
```

```python
# Plot x vs y
plt.figure(figsize=(8,5))
plt.plot(x, y, label='exp(x)', color='blue')

# Label axes and title
plt.xlabel("Time [milliseconds]")
plt.ylabel("Awesomeness")
plt.title("Exponential Awesomeness Growth")
plt.legend()

# Save the figure as a PDF file
plt.savefig("awesomeness_plot.pdf", format='pdf')

# Display the plot
plt.show()
```
