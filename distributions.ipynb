```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Generate 1000 random numbers from a normal (Gaussian) distribution
# mean = 0, standard deviation = 1
data = np.random.normal(loc=0, scale=1, size=1000)
```

```python
# Create a histogram with 100 bins
plt.figure(figsize=(8, 5))
plt.hist(data, bins=100, color='lightcoral', edgecolor='black')

# Label axes and title
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram of Normally Distributed Random Numbers")

# Save the figure as a PDF file
plt.savefig("normal_distribution_histogram.pdf", format='pdf')

# Display the plot
plt.show()
```
